```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
# Load dataset
df = pd.read_csv("/kaggle/input/datasets/farhankarim11/data-set/Position_Salary_Data.csv")

# Display first few rows
df.head()
```

```python
# Level as feature (X) and Salary as target variable (y)
X = df[['Level']].values
y = df['Salary'].values
```

```python
# Initialize and train linear regression model
model = LinearRegression()
model.fit(X, y)

# Predictions
y_pred = model.predict(X)

print(f"Slope (Coefficient): {model.coef_[0]:.2f}")
print(f"Intercept: {model.intercept_:.2f}")
```

```python
# Evaluate performance
r2 = r2_score(y, y_pred)
rmse = np.sqrt(mean_squared_error(y, y_pred))

print(f"R² Score: {r2:.4f}")
print(f"RMSE: {rmse:.2f}")
```

```python
# Plot dataset points and regression line
plt.figure(figsize=(8, 5))
plt.scatter(X, y, color='red', label='Actual Data')
plt.plot(X, y_pred, color='blue', linewidth=2, label='Linear Regression Line')
plt.title('Position Level vs Salary (Simple Linear Regression)')
plt.xlabel('Position Level')
plt.ylabel('Salary ($)')
plt.legend()
plt.grid(True)
plt.show()
```
